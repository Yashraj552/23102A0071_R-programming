The while loop, Example 2, a function that asks the user for a number and adds numbers until 25 is crossed

```r
sumfunction = function(){
  sum = 0
  number = as.integer(readline(prompt="Please select any number less than 25:   "))
  while (number <= 25) {
    sum = sum + number
    number = number + 1
  }
  print(paste("The sum of numbers received from the While Loop:   ", sum))
}
```

```r
i = 1
repeat{
  print( i^2 )
  i = i+2
  if ( i > 10 )
    break
}
```

```r
i = 1
repeat{
  i = i+1
  if (i < 10) next
  print(i^2)
  if (i >= 13) break
}
```
